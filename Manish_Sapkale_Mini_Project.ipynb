# Fill Missing Customer Names

```python
df["CustomerName"] = df["CustomerName"].fillna("Unknown")
```

```python
df["City"] = df["City"].fillna("Unknown")
```

```python
df.info()
```

```python
df["Quantity"].unique()
```

```python
df["UnitPrice"].unique()
```

```python
df["CustomerAge"].unique()
```

```python
df["Quantity"] = df["Quantity"].replace({
    "three": 3,
    "2 pcs": 2
})
```

```python
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")
```

```python
df["UnitPrice"] = df["UnitPrice"].str.replace(",", "", regex=False)
```

```python
df["UnitPrice"] = df["UnitPrice"].replace("unknown", pd.NA)
```

```python
df["UnitPrice"] = pd.to_numeric(df["UnitPrice"], errors="coerce")
```

```python
df["CustomerAge"] = df["CustomerAge"].replace("forty", 40)
```

```python
df["CustomerAge"] = df["CustomerAge"].str.replace(" years", "", regex=False)
```

```python
df["CustomerAge"] = pd.to_numeric(df["CustomerAge"], errors="coerce")
```

```python
df.info()
```

```python
df["Quantity"] = df["Quantity"].fillna(df["Quantity"].median())
```

```python
df["UnitPrice"] = df["UnitPrice"].fillna(df["UnitPrice"].median())
```

```python
df["CustomerAge"] = df["CustomerAge"].fillna(df["CustomerAge"].median())
```

```python
df["OrderDate"] = df["OrderDate"].fillna(df["OrderDate"].mode()[0])
```

```python
df["OrderStatus"] = df["OrderStatus"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["CustomerName"] = df["CustomerName"].str.strip()
df["City"] = df["City"].str.strip()
df["Product"] = df["Product"].str.strip()
df["OrderStatus"] = df["OrderStatus"].str.strip()
```

```python
df["City"] = df["City"].str.title()
```

```python
df["Product"] = df["Product"].str.title()
```

```python
df["OrderStatus"] = df["OrderStatus"].str.title()
```

```python
df["City"].unique()
```

```python
df["OrderDate"] = pd.to_datetime(df["OrderDate"], errors="coerce")
```

```python
df.info()
```

```python
df["OrderDate"] = df["OrderDate"].fillna(df["OrderDate"].mode()[0])
```

```python
df.info()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
import matplotlib.pyplot as plt

plt.boxplot(df["Quantity"])
plt.title("Quantity Outliers")
plt.show()
```

```python
plt.boxplot(df["CustomerAge"])
plt.title("CustomerAge Outliers")
plt.show()
```

```python
Q1 = df["Quantity"].quantile(0.25)
Q3 = df["Quantity"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df["Quantity"] >= lower) & (df["Quantity"] <= upper)]
```

```python
Q1 = df["CustomerAge"].quantile(0.25)
Q3 = df["CustomerAge"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df["CustomerAge"] >= lower) & (df["CustomerAge"] <= upper)]
```

```python
plt.boxplot(df["Quantity"])
plt.title("Quantity After Removing Outliers")
plt.show()
```

```python
plt.boxplot(df["CustomerAge"])
plt.title("CustomerAge After Removing Outliers")
plt.show()
```

```python
df["TotalAmount"] = df["Quantity"] * df["UnitPrice"]
```
